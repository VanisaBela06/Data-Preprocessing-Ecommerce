**Menangani Missing Value Numerik**

```python
# Menangani missing value numerik menggunakan mean

for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        mean_value = df[col].mean()

        print(f"{col}")
        print(f"Mean = {mean_value:.2f}")

        df[col] = df[col].fillna(mean_value)
```

```python
# Cek missing value numerik setelah imputasi

df[numeric_columns].isnull().sum()
```

**Menangani Missing Value Kategorik**

```python
# Menangani missing value kategorik menggunakan modus

for col in categorical_columns:
    if df[col].isnull().sum() > 0:

        mode_value = df[col].mode()[0]

        print(f"{col}")
        print(f"Modus = {mode_value}")

        df[col] = df[col].fillna(mode_value)
```

**Data Transformation — Min-Max Normalization**

```python
#Data Transformation - Min-Max Normalization

df['Pendapatan_Bulanan_Juta'] = pd.to_numeric(
    df['Pendapatan_Bulanan_Juta'],
    errors='coerce'
)

df['Rata_Rata_Transaksi_Ribu'] = pd.to_numeric(
    df['Rata_Rata_Transaksi_Ribu'],
    errors='coerce'
)

df['Total_Belanja_Ribu'] = pd.to_numeric(
    df['Total_Belanja_Ribu'],
    errors='coerce'
)

# Fungsi Min-Max Normalization
def min_max_normalization(series):
    return (series - series.min()) / (series.max() - series.min())

# Membuat kolom hasil normalisasi
df['Pendapatan_Normalized'] = min_max_normalization(
    df['Pendapatan_Bulanan_Juta']
)

df['Transaksi_Normalized'] = min_max_normalization(
    df['Rata_Rata_Transaksi_Ribu']
)

df['Total_Belanja_Normalized'] = min_max_normalization(
    df['Total_Belanja_Ribu']
)

print("Min-Max Normalization berhasil dilakukan.")

display(df[
    [
        'Pendapatan_Bulanan_Juta',
        'Pendapatan_Normalized',
        'Rata_Rata_Transaksi_Ribu',
        'Transaksi_Normalized',
        'Total_Belanja_Ribu',
        'Total_Belanja_Normalized'
    ]
].head(10))
```

```python
#Validasi hasil Min-Max Normalization

kolom_normalisasi = [
    'Pendapatan_Normalized',
    'Transaksi_Normalized',
    'Total_Belanja_Normalized'
]

for col in kolom_normalisasi:
    print(f"\n{col}")
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
```

**Discretization**

```python
#Discretization pada Usia

# Pastikan Usia numerik
df['Usia'] = pd.to_numeric(df['Usia'], errors='coerce')

# Membuat kategori usia
bins = [0, 17, 30, 45, 60, float('inf')]
labels = [
    'Remaja',
    'Dewasa Muda',
    'Dewasa',
    'Dewasa Akhir',
    'Lansia'
]

df['Kategori_Usia'] = pd.cut(
    df['Usia'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

print("Discretization Usia berhasil dilakukan.")

display(df[
    ['Usia', 'Kategori_Usia']
].head(15))
```
